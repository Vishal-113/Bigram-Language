### Given Confusion Matrix (System predictions as rows, Gold labels as columns)

| System \ Gold | **Cat** | **Dog** | **Rabbit** |
| ------------- | ------- | ------- | ---------- |
| **Cat**       | 5       | 10      | 5          |
| **Dog**       | 15      | 20      | 10         |
| **Rabbit**    | 0       | 15      | 10         |

Total = 90 animals.



## **Step 1. Per-Class Metrics**

**Formulas:**

* Precision for class *i*:
  [
  P_i = \frac{TP_i}{TP_i + FP_i}
  ]
* Recall for class *i*:
  [
  R_i = \frac{TP_i}{TP_i + FN_i}
  ]

Where:

* **TP** = true positives (correct predictions of class *i*)
* **FP** = predicted as *i* but actually another class
* **FN** = actual *i* but predicted as another class



### **Cat**

* TP = 5
* FP = (Predicted Cat but not Cat) = 10 + 5 = 15
* FN = (Gold Cat but not Cat) = 15 + 0 = 15
* Precision = ( 5 / (5+15) = 5/20 = 0.25 )
* Recall = ( 5 / (5+15) = 5/20 = 0.25 )



### **Dog**

* TP = 20
* FP = (Predicted Dog but not Dog) = 15 + 10 = 25
* FN = (Gold Dog but not Dog) = 10 + 15 = 25
* Precision = ( 20 / (20+25) = 20/45 ≈ 0.444 )
* Recall = ( 20 / (20+25) = 20/45 ≈ 0.444 )

---

### **Rabbit**

* TP = 10
* FP = (Predicted Rabbit but not Rabbit) = 0 + 15 = 15
* FN = (Gold Rabbit but not Rabbit) = 5 + 10 = 15
* Precision = ( 10 / (10+15) = 10/25 = 0.40 )
* Recall = ( 10 / (10+15) = 10/25 = 0.40 )



 **Per-class results:**

* Cat → Precision = 0.25, Recall = 0.25
* Dog → Precision ≈ 0.444, Recall ≈ 0.444
* Rabbit → Precision = 0.40, Recall = 0.40


## **Step 2. Macro vs Micro Averaging**

**Macro-average:** (simple mean across classes)

* Precision_macro = (0.25 + 0.444 + 0.40) / 3 ≈ 0.365
* Recall_macro = (0.25 + 0.444 + 0.40) / 3 ≈ 0.365


**Micro-average:** (aggregate TP, FP, FN across all classes)

* Total TP = 5 + 20 + 10 = 35

* Total FP = 15 + 25 + 15 = 55

* Total FN = 15 + 25 + 15 = 55

* Micro Precision = ( 35 / (35+55) = 35/90 ≈ 0.389 )

* Micro Recall = ( 35 / (35+55) = 35/90 ≈ 0.389 )

 **Final Results:**

* Macro Precision ≈ 0.365
* Macro Recall ≈ 0.365
* Micro Precision ≈ 0.389
* Micro Recall ≈ 0.389


**Interpretation:**

* **Macro averaging** treats all classes equally, regardless of frequency. (Useful when class balance matters).
* **Micro averaging** pools all decisions, so larger classes dominate. (Useful when overall performance across all samples matters).




```python
import numpy as np

# Confusion matrix (rows = predicted, cols = gold)
conf_matrix = np.array([
    [5, 10, 5],
    [15, 20, 10],
    [0, 15, 10]
])

# Number of classes
n_classes = conf_matrix.shape[0]

# Per-class precision and recall
precisions = []
recalls = []

for i in range(n_classes):
    TP = conf_matrix[i, i]
    FP = conf_matrix[i, :].sum() - TP
    FN = conf_matrix[:, i].sum() - TP

    precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0.0

    precisions.append(precision)
    recalls.append(recall)

# Macro averages
macro_precision = np.mean(precisions)
macro_recall = np.mean(recalls)

# Micro averages
TP_total = sum(conf_matrix[i, i] for i in range(n_classes))
FP_total = sum(conf_matrix[i, :].sum() - conf_matrix[i, i] for i in range(n_classes))
FN_total = sum(conf_matrix[:, i].sum() - conf_matrix[i, i] for i in range(n_classes))
micro_precision = TP_total / (TP_total + FP_total)
micro_recall = TP_total / (TP_total + FN_total)

# Print results
print("Per-Class Precision:", precisions)
print("Per-Class Recall:", recalls)
print("Macro Precision:", macro_precision)
print("Macro Recall:", macro_recall)
print("Micro Precision:", micro_precision)
print("Micro Recall:", micro_recall)
```
