```python
from collections import defaultdict

# Step 1: Training corpus
corpus = [
    ["<s>", "I", "love", "NLP", "</s>"],
    ["<s>", "I", "love", "deep", "learning", "</s>"],
    ["<s>", "deep", "learning", "is", "fun", "</s>"]
]

# Step 2: Unigram and Bigram counts
unigram_counts = defaultdict(int)
bigram_counts = defaultdict(int)

for sentence in corpus:
    for i in range(len(sentence)):
        unigram_counts[sentence[i]] += 1
        if i < len(sentence) - 1:
            bigram = (sentence[i], sentence[i+1])
            bigram_counts[bigram] += 1

# Vocabulary size
V = len(unigram_counts)

# Step 3: Estimate bigram probabilities (MLE)
def bigram_prob(w1, w2):
    return bigram_counts[(w1, w2)] / unigram_counts[w1] if unigram_counts[w1] > 0 else 0

# Step 4: Sentence probability function
def sentence_probability(sentence):
    prob = 1.0
    for i in range(len(sentence) - 1):
        p = bigram_prob(sentence[i], sentence[i+1])
        if p == 0:   # unseen bigram
            return 0
        prob *= p
    return prob

# Step 5: Test sentences
s1 = ["<s>", "I", "love", "NLP", "</s>"]
s2 = ["<s>", "I", "love", "deep", "learning", "</s>"]

p1 = sentence_probability(s1)
p2 = sentence_probability(s2)

print("Sentence 1:", " ".join(s1), "-> Probability:", p1)
print("Sentence 2:", " ".join(s2), "-> Probability:", p2)

# Step 6: Which sentence is preferred
if p1 > p2:
    print("\nModel prefers Sentence 1 because it has a higher probability.")
elif p2 > p1:
    print("\nModel prefers Sentence 2 because it has a higher probability.")
else:
    print("\nBoth sentences have equal probability.")
```
